## 1. Broadcasting in Action
Observe how a 1D bias vector of shape `(3,)` broadcasts across a 3D batch tensor of shape `(2, 4, 3)` without loop overhead.

```python
tokens = np.zeros((2, 4, 3))
bias = np.array([10.0, 20.0, 30.0])
result = tokens + bias
print("Tokens shape:", tokens.shape)
print("Bias shape:", bias.shape)
print("Broadcast Result shape:", result.shape)
print("Sample slice (batch 0, token 0):", result[0, 0])
```

## 2. Numerically Stable Softmax
Direct exponentiation $\exp(x)$ causes overflow when logits exceed $\sim 700$. Subtracting $\max(x)$ keeps the maximum exponent at $\exp(0) = 1$, eliminating overflow.

```python
def vectorized_softmax(logits: np.ndarray, axis: int = -1) -> np.ndarray:
    shift_logits = logits - np.max(logits, axis=axis, keepdims=True)
    exp_logits = np.exp(shift_logits)
    return exp_logits / np.sum(exp_logits, axis=axis, keepdims=True)

sample_logits = np.array([[1000.0, 1001.0, 1002.0], [1.0, 2.0, 3.0]])
probs = vectorized_softmax(sample_logits)
print("Probabilities:\n", probs)
print("Row sums (must equal 1.0):", probs.sum(axis=-1))
```

## 3. Vectorized Cosine Similarity Matrix
Pairwise cosine similarity between two matrices of embeddings `(N, D)` and `(M, D)` using matrix multiplication.

```python
def pairwise_cosine_similarity(A: np.ndarray, B: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    dot = A @ B.T
    norm_A = np.linalg.norm(A, axis=1, keepdims=True)
    norm_B = np.linalg.norm(B, axis=1, keepdims=True).T
    return dot / (norm_A * norm_B + eps)

np.random.seed(42)
embeddings_A = np.random.randn(5, 16)
embeddings_B = np.random.randn(8, 16)
sim_matrix = pairwise_cosine_similarity(embeddings_A, embeddings_B)

plt.figure(figsize=(6, 4), dpi=100)
plt.imshow(sim_matrix, cmap="coolwarm", vmin=-1.0, vmax=1.0)
plt.colorbar(label="Cosine Similarity")
plt.xlabel("Set B Vectors")
plt.ylabel("Set A Vectors")
plt.title("Pairwise Cosine Similarity Heatmap")
plt.tight_layout()
plt.show()
print("✅ Module 02 notebook executed successfully!")
```
