## 1. Type-Safe Experiment Configurations with `@dataclass`
Research code often suffers from loose dictionaries with typos (`cfg['lr']` vs `cfg['learning_rate']`). Using dataclasses provides auto-complete, validation, and serialization.

```python
@dataclass
class ExperimentConfig:
    model_name: str = "nano-transformer"
    learning_rate: float = 1e-3
    batch_size: int = 64
    max_steps: int = 1000
    seed: int = 42
    device: str = "cpu"

    def to_dict(self) -> Dict[str, Any]:
        return asdict(self)

    def save_json(self, filepath: str) -> None:
        with open(filepath, "w", encoding="utf-8") as f:
            json.dump(self.to_dict(), f, indent=4)
        print(f"✅ Configuration saved to {filepath}")

config = ExperimentConfig(learning_rate=3e-4, batch_size=32)
print("Experiment Config Object:", config)
config.save_json("config_example.json")
```

## 2. Memory-Efficient Batch Generator
Generators use the `yield` keyword to stream chunks of data one at a time, avoiding duplicate array allocations in RAM.

```python
def batch_generator(data: List[int], batch_size: int) -> Generator[List[int], None, None]:
    """Yields successive batches of items without allocating new full-length lists in memory."""
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

dataset = list(range(100))
print("=== Testing Batch Generator ===")
for step, batch in enumerate(batch_generator(dataset, batch_size=20)):
    print(f"Step {step}: Batch of size {len(batch)} -> {batch[:5]}...")

print("\n✅ Module 01 interactive notebook complete!")
```
