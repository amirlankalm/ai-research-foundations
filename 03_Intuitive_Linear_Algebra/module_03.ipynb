## 1. Visualizing Dot Products as Projections
The dot product $\vec{u} \cdot \vec{v} = \|\vec{u}\| \|\vec{v}\| \cos(\theta)$ measures how much two vectors point in the same direction.

```python
u = np.array([3.0, 1.0])
v = np.array([2.0, 2.0])

dot_val = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
cos_theta = dot_val / (norm_u * norm_v)
angle_deg = np.degrees(np.arccos(cos_theta))

print(f"Vector u: {u} (Norm: {norm_u:.2f})")
print(f"Vector v: {v} (Norm: {norm_v:.2f})")
print(f"Dot Product: {dot_val:.2f} | Cosine: {cos_theta:.3f} | Angle: {angle_deg:.1f}°")
```

## 2. SVD Low-Rank Approximation (The Core of LoRA)
Low-Rank Adaptation (LoRA) compresses massive weight matrices by decomposing them into low-rank factors. We verify how closely a rank-$k$ matrix approximates the full matrix.

```python
def svd_low_rank(W: np.ndarray, rank: int) -> np.ndarray:
    U, S, Vh = np.linalg.svd(W, full_matrices=False)
    return U[:, :rank] @ np.diag(S[:rank]) @ Vh[:rank, :]

W = np.random.randn(64, 64)
ranks = [1, 2, 4, 8, 16, 32, 64]
errors = []

for r in ranks:
    W_approx = svd_low_rank(W, rank=r)
    err = np.linalg.norm(W - W_approx, ord='fro')
    errors.append(err)
    print(f"Rank {r:2d} | Frobenius Error: {err:6.3f} ({err/np.linalg.norm(W):.1%} error)")

plt.figure(figsize=(7, 4), dpi=100)
plt.plot(ranks, errors, marker='o', lw=2, color='#805AD5')
plt.title("SVD Low-Rank Reconstruction Error", fontsize=13, fontweight='bold')
plt.xlabel("Rank k", fontsize=11)
plt.ylabel("Frobenius Error ||W - W_hat||", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
print("✅ Module 03 notebook executed successfully!")
```
